```python
import wave
import numpy as np
from pathlib import Path

# Use a WAV fallback to avoid requiring system ffmpeg for local transcription.
audio_path = Path("harvard.wav") if Path("harvard.wav").exists() else Path("harvard.mp3")

if audio_path.suffix.lower() == ".wav":
    with wave.open(str(audio_path), "rb") as wf:
        sample_rate = wf.getframerate()
        channels = wf.getnchannels()
        sample_width = wf.getsampwidth()
        n_frames = wf.getnframes()
        raw = wf.readframes(n_frames)

    # Convert PCM bytes to float32 in [-1, 1], then downmix to mono if needed.
    dtype_map = {1: np.int8, 2: np.int16, 4: np.int32}
    if sample_width not in dtype_map:
        raise ValueError(f"Unsupported WAV sample width: {sample_width}")

    audio = np.frombuffer(raw, dtype=dtype_map[sample_width]).astype(np.float32)
    if channels > 1:
        audio = audio.reshape(-1, channels).mean(axis=1)
    scale = float(np.iinfo(dtype_map[sample_width]).max)
    audio = audio / scale

    result = model.transcribe(audio)
else:
    result = model.transcribe(str(audio_path))

print(result["text"])
```
